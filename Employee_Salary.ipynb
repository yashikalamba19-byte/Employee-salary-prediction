**8**. **PERFORMANCE_SCORE**: Measures the employee's work performance rating.

**9**. **COMPANY_SIZE**: Describes the size of the organisation based on the number of employees.

**10**. **SALARY**(**TARGET** **VARIABLE**): Represents the monthly salary of the employee in INR.

**Encoding**

Here, we can see that **Age** is already a numerical, no encoding is required.

The next column is **Education Level**, on which we apply **Ordinal** **Encoding** since it follows hierarchical order.

```python
from sklearn.preprocessing import OrdinalEncoder
encoder1=OrdinalEncoder(categories=[["High School","Bachelor","Master","PhD"]])
q=encoder1.fit_transform(df[["Education_Level"]])
q
```

```python
df["Education_Level"]=df["Education_Level"].map({"High School":0,"Bachelor":1,"Master":2,"PhD":3})
```

The **Job Role** feature will be encoded using **One** **Hot** **Encoding** becuase it's best for Nominal Categorical Features

```python
from sklearn.preprocessing import OneHotEncoder
encoder2=OneHotEncoder()
w=encoder2.fit_transform(df[["Job_Role"]]).toarray()
w
```

```python
encoder2_df=pd.DataFrame(w,columns=encoder2.get_feature_names_out())
df=pd.concat([df,encoder2_df],axis=1)
df.head()
```

Now, the next column is **Department**.

we will apply **Label** **Encoding** here.

```python
from sklearn.preprocessing import LabelEncoder
encoder3=LabelEncoder()
df["Department"]=encoder3.fit_transform(df["Department"])
df.head()
```

Next column is **Location**.

we will apply **Label** **Encoding** here.

```python
from sklearn.preprocessing import LabelEncoder
encoder4=LabelEncoder()
df["Location"]=encoder4.fit_transform(df["Location"])
df.head()
```

Next column is **Certification**.

we will apply **Label** **Encoding** here.

```python
from sklearn.preprocessing import LabelEncoder
encoder5=LabelEncoder()
df["Certification"]=encoder5.fit_transform(df["Certification"])
df.head()
```

The last column is **Company_Size**,on which we will apply **Ordinal** **Encoding** beacuse it follows a hierarchical order.

```python
from sklearn.preprocessing import OrdinalEncoder
encoder6=OrdinalEncoder(categories=[["Small","Medium","Large"]])
e=encoder6.fit_transform(df[["Company_Size"]])
e
```

```python
df["Company_Size"]=df["Company_Size"].map({"Small":0,"Medium":1,"Large":2})
```

Now, we will drop categorical columns from the dataset

```python
df=df.drop(columns=["Job_Role"])
```

From here, we will separate **Independent** and **Dependent** variables

```python
x=df.drop(["Salary"],axis=1)
y=df["Salary"]
```

**Scaling** :

So, that all features are brought to the same scale.

```python
from sklearn.preprocessing import StandardScaler
scale=StandardScaler()
```

```python
x=scale.fit_transform(x)
```

Splitting DataSet into training and testing data

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=0)
```

**Choosing** **the** **best** **model**

**1**. **Linear Regression**

```python
from sklearn.linear_model import LinearRegression
regressor=LinearRegression()
```

```python
regressor.fit(x_train,y_train)
```

```python
y_pred=regressor.predict(x_test)
```

```python
y_pred
```

```python
y_test
```

Will calculate its metrics

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
```

```python
print(mean_squared_error(y_test,y_pred))
```

```python
print(mean_absolute_error(y_test,y_pred))
```

```python
print(r2_score(y_test,y_pred))
```

**2**. **Decission** **Tree** **Regressor**

```python
from sklearn.tree import DecisionTreeRegressor
dtr=DecisionTreeRegressor(max_depth=10)
```

```python
dtr.fit(x_train,y_train)
y_pred=dtr.predict(x_test)
print(r2_score(y_test,y_pred))
```

**3**. **Random** **Forest** **Regressor**

```python
from sklearn.ensemble import RandomForestRegressor
rfr=RandomForestRegressor(n_estimators=150,random_state=0)
```

```python
rfr.fit(x_train,y_train)
y_pred=rfr.predict(x_test)
print(r2_score(y_test,y_pred))
```

Here, we can see that **r2_score** is working well for our model selection.
